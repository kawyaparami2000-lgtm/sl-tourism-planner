## Sri Lanka Tourism Planner Multi-Agent Orchestration Test
This notebook verifies the multi-agent workflow consisting of:
- **Planner Agent** (Planning/Task-Decomposition Pattern)
- **Evaluator Agent** (Tool-Use Pattern with Stub Retriever)
- **Coach Agent** (Reflection/Self-Critique Pattern)
- **Router Node** (Orchestration & Edge Routing)
- **Model Router** (Sub-task Model Assignment: Groq for Evaluation, OpenRouter for Coaching)

```python
import json
from agents.graph import run_trip_planner
from models.model_router import get_model
```

```python
# Sample User Input for Phase 2 Prototyping
sample_input = {
    "travel_dates": "August, 7 days",
    "budget": "mid-range",
    "group_type": "couple",
    "interests": ["culture", "beaches"]
}

print("=== SAMPLE INPUT ===")
print(json.dumps(sample_input, indent=2))
```

```python
# Verify Model Assignments per Sub-Task
eval_model = get_model("evaluation")
coach_model = get_model("coaching_synthesis")
print(f"Evaluation Sub-Task Model: {eval_model}")
print(f"Coaching Synthesis Sub-Task Model: {coach_model}")
```

```python
# Execute Multi-Agent Graph Workflow
print("\n=== EXECUTING MULTI-AGENT TRIP PLANNER GRAPH ===")
final_state = run_trip_planner(sample_input)
```

```python
# Display Step-by-Step State Results
print("\n=== FINAL AGENT ORCHESTRATION STATE ===")

print("\n1. [PLANNER AGENT] Itinerary Legs (Planning / Task-Decomposition Pattern):")
for leg in final_state["itinerary_legs"]:
    print(f"  • Leg {leg['leg_number']}: {leg['destination']} ({leg['nights']} nights)")
    print(f"    Why it fits: {leg['why_it_fits']}")

print("\n2. [EVALUATOR AGENT] Retrieved Evidence & Feasibility (Tool-Use Pattern):")
print(f"   Model Assigned: {final_state['evaluation_notes'].get('model_used')}")
for fact in final_state["retrieved_evidence"]:
    print(f"  • Fact: {fact}")
print("   Feasibility Notes:", final_state["evaluation_notes"])

print("\n3. [COACH AGENT] Reflection & Optimization Critique (Reflection Pattern):")
print(final_state["coach_feedback"].get("formatted_summary"))
```
